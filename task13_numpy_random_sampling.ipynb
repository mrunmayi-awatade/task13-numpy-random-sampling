```python
# Generating random integers
import numpy as np
numbers = np.random.randint(1, 100, size=10)
print("Random Integers:", numbers)
```

```python
# Generating random floating-point numbers
import numpy as np
numbers = np.random.rand(5)
print("Random Float Numbers:", numbers)
```

```python
# Setting a random seed
import numpy as np
np.random.seed(42)
numbers = np.random.randint(1, 100, size=5)
print("Random Numbers:", numbers)
```

```python
# Generating reproducible random numbers
import numpy as np
np.random.seed(42)
numbers = np.random.rand(5)
print("Random Numbers:", numbers)
```

```python
# Selecting random samples from an array
import numpy as np
numbers = np.array([10, 20, 30, 40, 50, 60, 70, 80])
sample = np.random.choice(numbers, size=4, replace=False)
print("Random Sample:", sample)
```

```python
# Generating random numbers in a 2D array
import numpy as np
data = np.random.randint(1, 100, size=(3, 3))
print("Random 2D Array:")
print(data)
```

```python
# Shuffling array elements
import numpy as np
numbers = np.array([10, 20, 30, 40, 50])
np.random.shuffle(numbers)
print("Shuffled Array:", numbers)
```

```python
# Generating random student marks
import numpy as np
marks = np.random.randint(35, 101, size=10)
print("Student Marks:", marks)
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
# Creating a simulated sales dataset
import numpy as np
sales = np.random.randint(1000, 5000, size=7)
print("Daily Sales:", sales)
print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
```

```python
# Analyzing a random dataset
import numpy as np
data = np.random.randint(10, 100, size=10)
print("Dataset:", data)
print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))
```
